```python
import pandas as pd
import spotipy
import os
from spotipy.oauth2 import SpotifyClientCredentials
from dotenv import load_dotenv

load_dotenv(dotenv_path='../.env')

CLIENT_ID = os.getenv('SPOTIPY_CLIENT_ID')
CLIENT_SECRET = os.getenv('SPOTIPY_CLIENT_SECRET')

sp_oauth = spotipy.Spotify(auth_manager=SpotifyClientCredentials(client_id=CLIENT_ID, client_secret=CLIENT_SECRET))

raw_data_path = '../data/raw_data/sabrina_carpenter_discography.csv'
df = pd.read_csv(raw_data_path, sep=";")

last_id = df['track_id'].astype(int).max() if not df.empty else -1

album_id = 'https://open.spotify.com/album/1aqg30bNvLSWgShZgX4oop?si=qQFaOEINRX6eBXcdfkIUoQ'
album_info = sp_oauth.album(album_id)
tracks_result = sp_oauth.album_tracks(album_id)
tracks = tracks_result["items"]

new_songs = []

for track in tracks:
    last_id += 1
    #audio_features_list = sp_oauth.audio_features(track["id"])
    #audio_features = audio_features_list[0] if audio_features_list else None
    
    new_track = {
        'track_id': last_id,
        'track_name': track['name'],
        'track_type': 'B-Side',
        'track_musical_genre': '',   
        'duration_ms': track['duration_ms'],
        'spotify_streams': '',       
        'spotify_global_peak': '',   
        'spotify_usa_peak': '',      
        'track_videoclip': 'No',     
        'videoclip_views': 0,        
        'album': album_info['name'],
        'album_physical_sales': 0,   
        'track_number': track['track_number'],
        'album_musical_genre': '',   
        'release_date': album_info['release_date'],
        'uri': track['uri'],
        'acousticness': None,
        'danceability': None,
        'energy': None,
        'instrumentalness': None,
        'liveness': None,
        'loudness': None,
        'speechiness': None,
        'tempo': None,
        'valence': None   
    }
    
    new_songs.append(new_track)
    
df_new_songs = pd.DataFrame(new_songs)
df_updated = pd.concat([df, df_new_songs], ignore_index=True)
    
output_path = '../data/processed_data/new_sabrina_carpenter_discography.csv'
df_updated.to_csv(output_path, sep=";", index=False)
```
